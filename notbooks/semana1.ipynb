faça na linguagem python uma função que calcule as raizes da equação de segundo grau.

Serão passados os coeficientes a, b e c. Caso a = 0, resolva como uma equação do primeiro grau, e se b = 0 gere uma exeção.



## importando uma biblioteca
código usado para importar uma biblioteca especifica em python.

```python
from math import sqrt
```

## criando uma função
Raiz de equação quadratica.

```python
def zeros(a, b ,c):

    delta = b**2 - (4*a*c)

    if delta < 0:
        return None
    elif delta == 0:
        x = -b / (2*a)
        return [x, x]
    else:
        x1 = (-b + sqrt(delta)) / (2*a)
        x2 = (-b - sqrt(delta)) / (2*a)
        return [x1, x2]
```

## Executrando o código
Criando a função principal.

```python
def main():
    x = zeros(1, -5, 6)
    print(x)

    if __name__ == "__main__":
        main()
```
