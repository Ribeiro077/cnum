# Atividade 2
Faça na linguagem Python uma função que verifica se um número é perfeito Um número é perfeito se a soma dos divisores for igual a ele mesmo (ex: 6=1+2+3)


```python
def numero_perfeito(n):
    """Retorna 'perfeito' ou 'não perfeito'."""
    if n <= 0:
        return "não perfeito"

    soma_divisores = 1
    limite = int(n ** 0.5)

    for i in range(2, limite + 1):
        if n % i == 0:
            soma_divisores += i
            if i != n // i:
                soma_divisores += n // i

    if soma_divisores == n:
        return "perfeito"
    return "não perfeito"


# Exemplos
for numero in [6, 28, 15, 496, 10]:
    print(f"{numero} -> {numero_perfeito(numero)}")
```

6 -> perfeito
28 -> perfeito
15 -> não perfeito
496 -> perfeito
10 -> não perfeito

# Atividade 2

Faça na linguagem Python uma função que calcula o fatorial de um número.

O fatorial de um número é o produto do número por todos os antecessores positivos.

```python
def fatorial(n):
    """Retorna o fatorial de n."""
    if n < 0:
        return "Número inválido"
    if n == 0:
        return 1

    resultado = 1
    for i in range(1, n + 1):
        resultado *= i

    return resultado


# Exemplos
for numero in [0, 8, 5, 7]:
    print(f"{numero}! = {fatorial(numero)}")
```

0! = 1
1! = 1
5! = 120
7! = 5040
