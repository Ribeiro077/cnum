## Atividade 1
Faça na linguagem Python uma função que verifica se um número é perfeito.

Um número é perfeito se a soma dos divisores for igual a ele mesmo (ex: 6 = 1 + 2 + 3).

````python
def is_perfect(n: int) -> bool:
    if n < 1:
        return False

    sum_divisors = 0
    for i in range(1, n):
        if n % i == 0:
            sum_divisors += i

    return sum_divisors == n


def main():
    assert is_perfect(6) == True
    assert is_perfect(7) == False
    assert is_perfect(-1) == False

    print("Atividade 1")
    print("6 é perfeito?", is_perfect(6))
    print("7 é perfeito?", is_perfect(7))
    print("-1 é perfeito?", is_perfect(-1))


if __name__ == "__main__":
    main()
````

## Atividade 2
Faça na linguagem Python uma função que calcula o fatorial de um número.

O fatorial de um número é o produto do número por todos os antecessores positivos.

````python
def factorial(n: int) -> int:
    if n < 0:
        raise ValueError("O número deve ser não negativo.")

    result = 1
    i = n
    while i > 1:
        result *= i
        i -= 1

    return result

def main():
    assert factorial(5) == 120
    assert factorial(0) == 1
    try:
        factorial(-1)
    except ValueError as error:
        assert str(error) == "O número deve ser não negativo."

    print("Atividade 2")
    print("5! =", factorial(5))
    print("0! =", factorial(0))

if __name__ == "__main__":
    main()
````

## Atividade 3
Faça na linguagem Python uma função que verifica se um número é primo.

Um número que só pode ser dividido por um e por ele mesmo.

````python
def is_prime(n: int) -> bool:
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True


def main():
    assert is_prime(2) == True
    assert is_prime(3) == True
    assert is_prime(4) == False
    assert is_prime(1) == False
    assert is_prime(0) == False

    print("Atividade 3")
    print("2 é primo?", is_prime(2))
    print("3 é primo?", is_prime(3))
    print("4 é primo?", is_prime(4))
    print("1 é primo?", is_prime(1))
    print("0 é primo?", is_prime(0))


if __name__ == "__main__":
    main()
````

## Atividade 4
Faça na linguagem Python uma função que calcula a soma dos dígitos.

Caso número negativo, dispare uma mensagem de erro (ex: 123, 1 + 2 + 3 = 6).

````python
def sum_digits(n: int) -> int:
    if n < 0:
        raise ValueError("O número deve ser não negativo.")

    total = 0
    for digit in str(n):
        total += int(digit)
    return total


def main():
    assert sum_digits(123) == 6
    assert sum_digits(0) == 0
    try:
        sum_digits(-1)
    except ValueError as error:
        assert str(error) == "O número deve ser não negativo."

    print("Atividade 4")
    print("Soma dos dígitos de 123 =", sum_digits(123))
    print("Soma dos dígitos de 0 =", sum_digits(0))


if __name__ == "__main__":
    main()
````
