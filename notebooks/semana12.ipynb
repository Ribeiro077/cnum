## Atividade 1 — Integrais de a a f

```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
sys.path.append(str(Path.cwd().parent))
from scipy.integrate import quad
from utils.algoritmos import integral,medio,trapezio,simpson
fs=[("a",lambda x:np.exp(-x)),("b",lambda x:x**2),("c",lambda x:x**3),("d",lambda x:x*np.exp(-x**2)),("e",lambda x:1/(x*x+1)),("f",lambda x:x/(x*x+1))]
for nome,f in fs:
 vals=[integral(m,f,0,1,n=1000) for m in (medio,trapezio,simpson)]
 print(nome,*[round(v,8) for v in vals],"SciPy",round(quad(f,0,1)[0],8))
```

## Atividade 2 — Integral composta de 2 a 5
