## Atividade 5 — Corrente linear

```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
sys.path.append(str(Path.cwd().parent))
from utils.algoritmos import regressao,dp
t=np.array([0,.1,.2,.3,.4]); I=np.array([0,.82,1.36,1.60,1.73])
a=regressao(t,I,lambda z:np.column_stack((np.ones_like(z),z))); f=lambda z:a[0]+a[1]*z
d=dp(f,.5,.1); print("coeficientes",a,"dI/dt",d,"V(0.5)",.1*d+5*f(.5))
```

## Atividade 6 — Corrente quadrática

```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
sys.path.append(str(Path.cwd().parent))
from utils.algoritmos import regressao,dr
t=np.array([0,.1,.2,.3,.4]); I=np.array([0,.82,1.36,1.60,1.73])
a=regressao(t,I,lambda z:np.column_stack((z**2,z,np.ones_like(z)))); f=lambda z:a[0]*z*z+a[1]*z+a[2]
d=dr(f,.5,.1); print("coeficientes",a,"dI/dt",d,"V(0.5)",.1*d+5*f(.5))
```

## Atividade 7 — Corrente cúbica

```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
sys.path.append(str(Path.cwd().parent))
from utils.algoritmos import regressao,dc
t=np.array([0,.1,.2,.3,.4]); I=np.array([0,.82,1.36,1.60,1.73])
a=regressao(t,I,lambda z:np.column_stack((z**3,z**2,z,np.ones_like(z)))); f=lambda z:((a[0]*z+a[1])*z+a[2])*z+a[3]
d=dc(f,.5,.1); print("coeficientes",a,"dI/dt",d,"V(0.5)",.1*d+5*f(.5))
```
