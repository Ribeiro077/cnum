## Atividade 4 — Ajustes trigonométrico e cúbico

```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
sys.path.append(str(Path.cwd().parent))
from utils.algoritmos import regressao
x=np.arange(0,1.01,.1); y=np.array([31,35,37,33,28,20,16,15,18,23,31.])
a=regressao(x,y,lambda z:np.column_stack((np.ones_like(z),np.sin(2*np.pi*z),np.cos(2*np.pi*z))))
b=regressao(x,y,lambda z:np.column_stack((np.ones_like(z),z,z**2,z**3)))
print("a,b,c trigonométrico =",a); print("a,b,c,d cúbico =",b)
xx=np.linspace(0,1,400); plt.scatter(x,y); plt.plot(xx,np.column_stack((np.ones_like(xx),np.sin(2*np.pi*xx),np.cos(2*np.pi*xx)))@a,label="Trig"); plt.plot(xx,np.column_stack((np.ones_like(xx),xx,xx**2,xx**3))@b,label="Cúbico"); plt.legend(); plt.grid(); plt.show()
```
